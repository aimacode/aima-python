# Search in the browser

Uninformed and informed search on the Romania map from chapter 3, running
fully in your browser with `aima.search`. Run the cells top to bottom.

```python
import piplite
await piplite.install("aima", deps=False)
```

```python
from aima.search import GraphProblem, romania_map
from aima.search import breadth_first_graph_search, uniform_cost_search, astar_search

problem = GraphProblem("Arad", "Bucharest", romania_map)

for name, algorithm in [
    ("Breadth-first", breadth_first_graph_search),
    ("Uniform-cost", uniform_cost_search),
    ("A* (straight-line heuristic)", astar_search),
]:
    goal = algorithm(problem)
    print(f"{name:30s} {['Arad'] + goal.solution()}  (cost {goal.path_cost})")
```

A* finds the optimal route Arad → Sibiu → Rimnicu → Pitesti → Bucharest
(cost 418), while breadth-first returns a shorter-in-hops but costlier path.

## Watching the search

The players below replay a run of the same search functions, one expansion per
step: press Play or Step, or drag the slider. Yellow cities are in the frontier,
blue ones are explored and the red one is being expanded; the last step draws the
path found and its cost.

```python
await piplite.install("ipywidgets")
from aima.widgets import graph_search_player

graph_search_player(astar_search, problem)
```

Breadth-first search ignores the road lengths: it reaches Bucharest in fewer
steps, by a longer road.

```python
graph_search_player(breadth_first_graph_search, problem)
```

Uniform-cost search expands the cities in order of their distance from Arad, so
it expands 12 cities before it finds the route that A* finds after 5.

```python
graph_search_player(uniform_cost_search, problem)
```

The players are built on `RecordingProblem`, which wraps a problem and records
what the search asks of it; the search function is unchanged and returns the same
node.

```python
from aima.search import RecordingProblem

recorder = RecordingProblem(problem)
goal = astar_search(recorder)
[step["current"] for step in recorder.steps(goal)]
```

## The 8-puzzle

A* with the misplaced-tiles heuristic (`EightPuzzle.h`) solves this puzzle; the
player shows the moves of the solution it finds.

```python
from aima.search import EightPuzzle
from aima.widgets import eight_puzzle_player, EightPuzzleGame

eight_puzzle_player(astar_search, EightPuzzle((1, 2, 0, 6, 8, 3, 5, 4, 7)))
```

Your turn: click a tile next to the blank to slide it. When the puzzle is solved,
your number of moves is compared with the optimum that A* finds.

## The n-queens problem

`NQueensProblem` fills the columns from left to right and places a queen only
where no other queen attacks it. Depth-first search backtracks when a column has no
safe square; the highlighted column is the next one to fill.

```python
from aima.search import NQueensProblem, depth_first_tree_search
from aima.widgets import n_queens_player

n_queens_player(depth_first_tree_search, NQueensProblem(6))
```
