## Propositional logic

Model checking by truth-table enumeration, and satisfiability via DPLL.

```python
from aima.logic import expr, tt_entails, dpll_satisfiable

# modus ponens: does  P & (P => Q)  entail  Q ?
print("(P & (P ==> Q)) |= Q :", tt_entails(expr("(P & (P ==> Q))"), expr("Q")))

# satisfiability
print("P | ~P satisfiable   :", bool(dpll_satisfiable(expr("P | ~P"))))
print("P & ~P satisfiable   :", bool(dpll_satisfiable(expr("P & ~P"))))
```

## First-order logic

The classic "West is a criminal" knowledge base (section 9.3), solved by
forward chaining.

```python
from aima.logic import expr, FolKB, fol_fc_ask

clauses = [
    "(American(x) & Weapon(y) & Sells(x, y, z) & Hostile(z)) ==> Criminal(x)",
    "Owns(Nono, M1)",
    "Missile(M1)",
    "(Missile(x) & Owns(Nono, x)) ==> Sells(West, x, Nono)",
    "Missile(x) ==> Weapon(x)",
    "Enemy(x, America) ==> Hostile(x)",
    "American(West)",
    "Enemy(Nono, America)",
]
kb = FolKB([expr(c) for c in clauses])
print("Who is a criminal? ", list(fol_fc_ask(kb, expr("Criminal(x)"))))
```

## Watching DPLL

The wumpus KB of section 7.4.3 says that there is no breeze in [1,1] and a breeze
in [2,1]. To prove that there is no pit in [1,2], DPLL shows that the KB together
with P12 has no model. The player replays each call of DPLL: the model built so
far, why the last symbol got its value (pure symbol, unit clause or branch), and
each clause of the CNF in green when true and in red when false.

```python
await piplite.install("ipywidgets")
from aima.logic import wumpus_kb, associate
from aima.widgets import dpll_player, WumpusWorldGame

dpll_player(associate("&", wumpus_kb.clauses) & expr("P12"))
```

A satisfiable sentence ends with a model instead:

```python
dpll_player(expr("(A | B) & (~A | C) & (~C | ~B)"))
```
