## Map colouring

Colour the map of Australia with three colours so that no two neighbouring
regions share one. First we enforce arc consistency (AC-3), then solve by
backtracking search.

```python
from aima.csp import australia_csp, AC3, backtracking_search

consistent, checks = AC3(australia_csp)
print("AC-3 arc-consistent:", consistent, "  (", checks, "constraint checks )")
print("backtracking solution:", backtracking_search(australia_csp))
```

## N-queens

Place 8 non-attacking queens by local search (min-conflicts). The solution
maps each column to the row of its queen.

```python
import random
from aima.csp import NQueensCSP, min_conflicts

random.seed(0)
solution = min_conflicts(NQueensCSP(8), max_steps=1000)
print("8-queens (column -> row):", solution)
```

## Watching the solvers

The players below replay a run of the same solvers, one assignment, backtrack or
pruning per step: press Play or Step, or drag the slider. A region is filled with
its color once assigned, and the small squares under it are the values left in its
domain.

```python
await piplite.install("ipywidgets")
from aima.csp import MapColoringCSP, forward_checking
from aima.widgets import csp_player, MapColoringGame


def australia():
    return MapColoringCSP(list("RGB"), "SA: WA NT Q NSW V; NT: WA Q; NSW: Q V; T: ")


csp_player(backtracking_search, australia(), inference=forward_checking)
```

AC-3 can find that a partial assignment has no solution before any search. With
WA = R and Q = G, it prunes the domains until a region has no value left.

```python
def ac3_after_wa_red_and_q_green(csp):
    removals = csp.suppose("WA", "R") + csp.suppose("Q", "G")
    return AC3(csp, removals=removals)


csp_player(ac3_after_wa_red_and_q_green, australia())
```

Min-conflicts starts from a complete assignment and moves one queen at a time to
the row with fewest conflicts; queens under attack are red.

```python
random.seed(0)
csp_player(min_conflicts, NQueensCSP(8), max_steps=1000)
```
