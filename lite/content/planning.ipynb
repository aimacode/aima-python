# Classical planning in the browser

STRIPS-style planning (chapter 11) solved with GraphPlan, running fully in
your browser with `aima.planning`.

```python
import piplite
await piplite.install("aima", deps=False)
```

## Spare tire

Replace a flat tire with the spare from the trunk. `GraphPlan` returns a
layered plan; `linearize` flattens it into an executable action sequence.

```python
from aima.planning import spare_tire, air_cargo, GraphPlan, linearize

plan = GraphPlan(spare_tire()).execute()
print("spare tire plan:", linearize(plan))
```

## Air cargo

Fly two cargos to their destinations (Figure 11.2) — a larger problem with
load/fly/unload actions across two airports.

```python
plan = GraphPlan(air_cargo()).execute()
for step, action in enumerate(linearize(plan), 1):
    print(f"{step}. {action}")
```

## Watching the plans

The player below executes the plan GraphPlan finds for the three-block tower
(Figure 10.3, the Sussman anomaly), one action per step: press Play or Step, or
drag the slider. The block that moves is highlighted.

```python
await piplite.install("ipywidgets")
from aima.planning import three_block_tower
from aima.widgets import plan_player, graph_plan_player

plan_player(three_block_tower(), linearize(GraphPlan(three_block_tower()).execute()))
```

For a problem that is not a blocks world the player lists the facts of each
state: the ones the action adds are in green, the ones it deletes are struck
through, and the goals that hold are checked.

```python
plan_player(spare_tire(), linearize(GraphPlan(spare_tire()).execute()))
```

GraphPlan builds a planning graph level by level until the goals hold with no
two of them mutex, then extracts the plan backwards. The player shows each level:
the literals of the state level, the actions of the action level (the persistence
actions are only counted) and the number of mutex pairs.

```python
graph_plan_player(spare_tire())
```
