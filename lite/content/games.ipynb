# Adversarial search in the browser

Minimax and alpha-beta on Tic-Tac-Toe from chapter 5, running fully in your
browser with `aima.games`.

```python
import piplite
await piplite.install("aima", deps=False)
```

```python
from aima.games import TicTacToe, minmax_decision, alpha_beta_search

game = TicTacToe()
state = game.initial
print("minimax optimal first move:   ", minmax_decision(state, game))
print("alpha-beta optimal first move:", alpha_beta_search(state, game))
```

```python
import random
from aima.games import alpha_beta_player, random_player

# An optimal alpha-beta player never loses: against a random opponent it wins
# or draws every game (utility >= 0 for the first player).
random.seed(0)
results = [game.play_game(alpha_beta_player, random_player) for _ in range(10)]
print("alpha-beta vs random, 10 games (first-player utility):", results)
print("never loses:", all(u >= 0 for u in results))
```

## Watching the game tree

The player below replays the tree that minimax explores on the two-ply game of
Figure 5.2, one node per step: MAX nodes are up triangles, MIN nodes down
triangles and leaves squares. A node shows its backed-up value once its subtree is
done. Press Play or Step, or drag the slider.

```python
await piplite.install("ipywidgets")
from aima.games import Fig52Game
from aima.widgets import game_tree_player, TicTacToeGame

game_tree_player(minmax_decision, Fig52Game())
```

Alpha-beta returns the same move but skips part of the tree: after C1 = 2, MIN
can hold C to 2 or less, while MAX already has 3 from B, so C2 and C3 are pruned.

```python
game_tree_player(alpha_beta_search, Fig52Game())
```
